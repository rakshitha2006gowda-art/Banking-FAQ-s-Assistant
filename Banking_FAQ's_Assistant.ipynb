#Context Engineering
## Objective

In this lab, we will learn how to:

Structure context instead of dumping prompts
Separate static, dynamic, and memory context
Assemble context using Python
Call an AI model with structured context
Build a real-world Banking FAQ Assistant

# Use Case
Banking FAQ Assistant

Key idea: **Context is infrastructure, not text.**

### **Problem Statement**
We want to build an AI system that:

   Acts as a banking customer support agent

   Follows bank policies and FAQs strictly

   Responds politely and clearly

   Uses customer-specific information

   Uses previous interaction information

   Escalates sensitive issues when required

This is not a generic chatbot.

This is a context-aware banking AI system.

## Context Engineering Approach

We break context into structured parts:

1. Static Context → Who the AI is and how it behaves
2. External Context → Policies or documents
3. Dynamic Context → User input and task data
4. Memory Context → What the system remembers

We assemble these intentionally.

# **Step 1: Static Context**
Defines the role of the AI.

Specifies banking rules and guidelines.

Maintains a professional and polite tone.

Ensures answers are clear and simple.

Prevents the AI from inventing banking information.

Protects sensitive information such as OTP, PIN, CVV, and passwords.

Defines when to escalate issues to human support.

Usually remains unchanged for different customers.

```python
SYSTEM_CONTEXT = """
You are a Banking FAQ Assistant for a bank.

Rules:
- Be polite, clear and professional.
- Answer using the provided banking information.
- Do not invent banking policies or fees.
- Never ask for or expose PIN, OTP, CVV or passwords.
- If the information is unavailable, say that you do not know.
- For sensitive account issues, recommend contacting the bank securely.
"""
```

## **Step 2: External Context (Banking FAQs & Policies)**

The AI should not guess banking information.

We explicitly provide bank FAQs, rules, and policies as context.

```python
BANKING_POLICY = """
Banking FAQ Policy:

- ATM card replacement fee is ₹200.
- Savings account minimum balance is ₹5,000.
- IMPS is available 24/7.
- UPI transactions are subject to bank and network availability.
- Customers should never share their OTP, PIN or CVV.
- For a lost ATM card, the customer should immediately block the card
  through the official banking channel.
"""
```

# **Step 3: Dynamic Context**

Dynamic context changes per customer request.

This includes the current user question and customer-specific information.


```python
user_query = "I lost my ATM card. What should I do?"

user_profile = {
    "customer_type": "Savings Account Customer",
    "preferred_language": "English",
    "card_status": "Active"
}
```

## **step 4: Memory Context**

Memory stores useful information from previous interactions.

```python
SESSION_MEMORY = """
User previously asked about ATM card services.
User prefers short and simple explanations.
"""
```

## **Step 5: Context Assembly**

We now combine the static, external, dynamic, and memory context carefully.

Order and clarity matter because the AI needs the right information at the right time.

This is context engineering.

```python
final_prompt = f"""
{SYSTEM_CONTEXT}

BANKING POLICY:
{BANKING_POLICY}

USER PROFILE:
- Customer Type: {user_profile['customer_type']}
- Preferred Language: {user_profile['preferred_language']}
- Card Status: {user_profile['card_status']}

SESSION MEMORY:
{SESSION_MEMORY}

CURRENT USER QUESTION:
{user_query}
"""

print(final_prompt)
```

## Step 6: Call the Model

We now send the structured context to the model.

```python
from openai import OpenAI

client = OpenAI(
    api_key=api_key,
    base_url="https://nexusapi.navigatelabs.ai"
)

response = client.chat.completions.create(
    model="gemini-2.5-flash",
    messages=[
        {
            "role": "system",
            "content": SYSTEM_CONTEXT
        },
        {
            "role": "user",
            "content": final_prompt
        }
    ]
)

print(response.choices[0].message.content)
```
