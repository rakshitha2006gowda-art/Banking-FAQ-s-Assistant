**Introduction**
An API (Application Programming Interface) allows one program to request data or services from another program over the internet.

The workflow follows this pattern:

Python --> Request --> API --> JSON Response --> Process --> Display

You send a request, the API processes it and sends back data (usually in JSON format), and your program reads and uses that data.

**Step 1 -- Install / Import Requests**

The requests library makes it easy to send HTTP requests from Python.

If the library is not already installed, run the following command first:

```python
# Install requests (only needed once -- skip if already installed)
# Uncomment the line below if you get an ImportError
# !pip install requests
```

```python
# Import the requests library
import requests
```

If the import runs without errors, the library is ready to use.

**Step 2 -- Define the API URL**

Every API has a specific URL (called an endpoint) that you send your request to.

We will request information about user number 1 from JSONPlaceholder.

```python
# Define the API endpoint
url = "https://jsonplaceholder.typicode.com/users/1"
```

This URL points to a specific resource: user with ID 1. Changing the number at the end (e.g., /users/2) would request a different user.

**Step 3 -- Send the Request**

requests.get() sends an HTTP GET request to the API and stores the response.

```python
# Send a GET request to the API
response = requests.get(url)

# Check the status code
print("Status Code:", response.status_code)
```

**Step 4 -- Convert Response to JSON**

APIs return data in **JSON** (JavaScript Object Notation) format. In Python, JSON data looks and works like a dictionary.

```python
# Convert the response to a Python dictionary
data = response.json()

# Display the full response
print(data)
```

**Step 5 -- Extract Selected Values**

You rarely need all the data from an API. Use dictionary keys to extract the specific fields you want.

```python
# Extract specific fields from the response
print("Name :", data["name"])
print("Email:", data["email"])
print("City :", data["address"]["city"])
```

**Step 6 -- Convert the API Call into a Function**

To make the API logic reusable, wrap it inside a function. This way you can fetch data for any URL with a single function call.

```python
def get_api_data(url):
    """Send a GET request to the given URL and return JSON data.
    Returns None if the request fails."""
    response = requests.get(url)

    if response.status_code == 200:
        return response.json()

    return None
```

```python
# Use the function to fetch user 1
user_data = get_api_data("https://jsonplaceholder.typicode.com/users/1")

if user_data is not None:
    print("Name :", user_data["name"])
    print("Email:", user_data["email"])
    print("City :", user_data["address"]["city"])
else:
    print("Failed to fetch data from the API.")
```

```python
# Demonstrate reuse -- fetch a different user
user_2 = get_api_data("https://jsonplaceholder.typicode.com/users/2")

if user_2 is not None:
    print("Name :", user_2["name"])
    print("Email:", user_2["email"])
    print("City :", user_2["address"]["city"])
else:
    print("Failed to fetch data from the API.")
```

The same function works for any user endpoint -- you just change the URL.

**Description**

The code sends a GET request to the API, checks whether the request is successful, converts the response into JSON data, and extracts selected user details such as name, username, email, phone, and company. The API logic is then placed inside reusable functions for easier use with different users.

```python
import requests

def get_api_data(url):
    response = requests.get(url)

    if response.status_code == 200:
        return response.json()
    return None

def display_user_info(data):
    print("\n--- USER INFORMATION ---")
    print("Name    :", data["name"])
    print("Username:", data["username"])
    print("Email   :", data["email"])
    print("Phone   :", data["phone"])
    print("Company :", data["company"]["name"])

data = get_api_data(
    "https://jsonplaceholder.typicode.com/users/5"
)

if data:
    display_user_info(data)
else:
    print("Failed to fetch data.")
```
